# Data Analytics


**Data Analytics** é a área que **analisa dados para extrair informações úteis, identificar padrões e apoiar a tomada de decisão**.

Em outras palavras, é o processo de **transformar dados em conhecimento**.



## Definição simples

**Data Analytics = coleta + organização + análise + interpretação de dados**

O objetivo é responder perguntas como:

- O que aconteceu?
- Por que aconteceu?
- O que pode acontecer no futuro?
- O que devemos fazer com base nesses dados?



## Tipos principais de Data Analytics

Costuma-se dividir em quatro tipos:

### 1. Análise Descritiva (Descriptive Analytics)

Responde **o que aconteceu**.

Exemplos:

- Média de vendas do mês  
- Número de usuários de um aplicativo  
- Taxa de aprovação de alunos  

Ferramentas comuns:

- gráficos  
- tabelas  
- dashboards  



### 2. Análise Diagnóstica (Diagnostic Analytics)

Responde **por que aconteceu**.

Exemplos:

- Por que as vendas caíram?  
- Qual fator está associado à evasão escolar?

Ferramentas:

- correlação  
- análise exploratória de dados (AED)  
- regressão  



### 3. Análise Preditiva (Predictive Analytics)

Responde **o que provavelmente acontecerá**.

Usa **modelos estatísticos e machine learning**.

Exemplos:

- prever inadimplência  
- prever demanda  
- prever churn de clientes  

Modelos comuns:

- regressão  
- árvores de decisão  
- redes neurais  
- ARIMA  



### 4. Análise Prescritiva (Prescriptive Analytics)

Responde **o que devemos fazer**.

Envolve **otimização e decisão**.

Exemplos:

- qual preço maximiza lucro  
- qual rota minimiza custo  
- qual estoque manter

## Um pouco mais sobre Análise Descritiva e Análise Preditiva

### 1. Análise Descritiva
A **Análise Descritiva** tem como objetivo principal **resumir os dados para extrair padrões** de eventos que já ocorreram. Nessa categoria, os resultados são obtidos diretamente pela aplicação de algoritmos aos dados brutos, gerando estatísticas, visualizações ou agrupamentos. As principais técnicas descritivas incluem:

*   **Estatística Descritiva:** Utilizada para descrever conjuntos de dados (amostras) por meio de sumarização e visualização, podendo ser univariada (uma variável ou atributo), bivariada (pares de variáveis ou atributos) ou multivariada (mais de duas variáveis ou atributos).

*   **Visualização de Dados:** Usa gráficos como histogramas, box-plots, gráficos de dispersão e heatmaps para fornecer insights visuais sobre a distribuição e relação dos dados.

*   **Agrupamento (Clustering):** Técnica que visa encontrar **agrupamentos de dados (grupos de instâncias) semelhantes** dentro de um conjunto de dados, sendo amplamente aplicada em áreas como o marketing para segmentação de clientes.

*   **Mineração de Padrões Frequentes:** Focada em capturar as associações e padrões que ocorrem com maior regularidade nos dados, como na análise de cestas de compras no mercado varejista.


### 2. Análise Preditiva

A **Análise Preditiva** foca na **extração de modelos a partir dos dados para realizar previsões** sobre o desconhecido ou o futuro. Ao contrário da descritiva, ela utiliza dados rotulados (treinamento) para induzir um modelo que possa associar corretamente os atributos de novos objetos a um rótulo de saída. As tarefas preditivas são subdivididas em:

*   **Regressão:** Envolve a **predição de um valor quantitativo** (numérico), como o preço de um ativo, o tempo de viagem ou indicadores macroeconômicos.

*   **Classificação:** O objetivo é atribuir uma categoria ou **rótulo qualitativo** (classe) a um novo objeto, como decidir se um e-mail é spam ou não, diagnosticar se um paciente está doente ou prever se uma transação financeira é fraudulenta.

Em resumo, enquanto a tecnologia de Big Data foca na coleta e gerenciamento (volume, variedade e velocidade), o Data Analytics e a Ciência de Dados utilizam essa taxonomia para **identificar padrões e comportamentos úteis nos dados**.

## Principais diferenças entre C e Python

C_Python.png



As linguagens **C** e **Python** possuem propósitos e filosofias diferentes de desenvolvimento.

**1. Nível da linguagem**
- **C** é uma linguagem de **baixo/médio nível**, mais próxima do hardware.
- **Python** é uma linguagem de **alto nível**, focada em simplicidade e produtividade.

**2. Compilação vs interpretação**
- **C** é **compilada**: o código precisa ser compilado antes da execução.
- **Python** é **interpretada**: o código é executado diretamente pelo interpretador.

**3. Tipagem**
- **C** possui **tipagem estática**: o tipo da variável deve ser declarado.
- **Python** possui **tipagem dinâmica**: o tipo é definido automaticamente durante a execução.

**4. Gerenciamento de memória**
- Em **C**, o programador precisa gerenciar a memória manualmente (malloc, free).
- Em **Python**, o gerenciamento de memória é **automático** (garbage collection).

**5. Sintaxe**
- **C** possui sintaxe mais rígida, com uso obrigatório de `{}`, `;` e declaração explícita de tipos.
- **Python** possui sintaxe mais simples e legível, utilizando **indentação** para definir blocos.

**6. Aplicações típicas**
- **C**: sistemas operacionais, drivers, sistemas embarcados e aplicações de alto desempenho.
- **Python**: ciência de dados, inteligência artificial, automação, web e prototipagem rápida.

Em resumo, **C oferece maior controle e desempenho**, enquanto **Python prioriza simplicidade e rapidez no desenvolvimento**.

## Boas práticas (PEP 8)

Python Enhancement Proposals:
PEP 8 Style Guide for Python Code: https://peps.python.org/pep-0008/

Recomendações dos slides:
- Nomes descritivos: `media_alunos` em vez de `m`
- Convenção:
  - variáveis e funções: `minusculo_com_underscore`
  - classes: `CamelCase`
  - constantes: `MAIUSCULO_COM_UNDERSCORE`

```python
# Exemplos de convenções

# constante
TAXA_MAXIMA = 0.25

# função
def calcula_media(valores):
    return sum(valores)/len(valores)

# classe
class Pessoa:
    def __init__(self, nome):
        self.nome = nome
```

```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome      # atributo de instância
        self.idade = idade

    def apresentar(self):     # método de instância
        return f"Olá, meu nome é {self.nome} e tenho {self.idade} anos."

# Criando objetos (instâncias)
p1 = Pessoa("Ezequias", 30)
p2 = Pessoa("Maria", 25)

print(p1.apresentar())  # Olá, meu nome é Ezequias e tenho 30 anos.
print(p2.apresentar())  # Olá, meu nome é Maria e tenho 25 anos.
```

```python
# Exemplo do uso da função calcula_media

dados = [10, 8, 9]
media = calcula_media(dados)

print(media)
```

```python
# Exemplo do uso da classe Pessoa

p = Pessoa("Ana")

# print(p.nome)
print(p.nome)
```

```python
# Outro exemplo de classe

class Carro:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo
```

```python
# Outro exemplo de classe

c1 = Carro("Toyota", "Corolla")

print(c1.marca)
print(c1.modelo)
```

O que é uma classe, afinal?

Uma classe em Python é um modelo (ou molde) usado para criar objetos.

Ela define quais dados (atributos) e quais ações (métodos) os objetos daquele tipo terão.

Uma classe é uma estrutura que define um tipo de objeto, especificando seus atributos e métodos.


```python
# E com método = ações que a classe pode executar

class Pessoa:
    def __init__(self, nome):
        self.nome = nome

    def cumprimentar(self):
        print("Olá, eu sou", self.nome)

p = Pessoa("João")
p.cumprimentar()
```

## 7) Operadores
### 7.1) Aritméticos
- `+`, `-`, `*`, `/`, `**`, `//`, `%`

### 7.2) Relacionais
- `==`, `!=`, `>`, `>=`, `<`, `<=`

### 7.3) Lógicos
- `not`, `and`, `or`

```python
# Aritméticos
a = 10
b = 3

print('a=',a)
print('b=',b)
```

```python
print("a+b =", a+b) # soma
print("a-b =", a-b) # subtração
print("a*b =", a*b) # produto
print("a/b =", a/b) # divisão
print("a**b =", a**b) # potenciação
print("a//b =", a//b) # divisão inteira
print("a%b =", a%b) # mod = resto da divisão
```

```python
# Relacionais
print("a == b?", a == b) # testa igualdade
print("a != b?", a != b) # testa diferença
print("a >= b?", a >= b)
```

```python

# Operadores lógicos (combinam condições booleanas)

print("not (a > b)?", not (a > b))
# negação lógica: inverte o resultado da condição (True vira False e vice-versa)

print("(a > b) and (b > 0)?", (a > b) and (b > 0))
# AND lógico: retorna True apenas se ambas as condições forem verdadeiras

print("(a < b) or (b > 0)?", (a < b) or (b > 0))
# OR lógico: retorna True se pelo menos uma das condições for verdadeira
```
